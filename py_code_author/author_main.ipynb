```python
# Cell 1: Imports and Params
import pandas as pd
from preprocessing import run_preprocessing
from analysis import run_markov_analysis, run_decoding_analysis
from author_statistics import calculate_bfs
from visualization import plot_behavior, plot_decoding

PATH_DATA = '../project/ds006761/'
# Exclude pairs 10, 23, 24 due to data quality issues [cite: 498]
#PAIR_IDS = [i for i in range(1, 35) if i not in [10, 23, 24]]
PAIR_IDS = [i for i in range(1, 2) if i not in [10, 23, 24]]
participants = pd.read_csv('../project/ds006761/participants.tsv', sep='\t')
```

```python
# Cell 2: Pipeline Execution
print("Step 1: Preprocessing EEG...")
epochs_dict = run_preprocessing(PATH_DATA, PAIR_IDS, participants)

print("Step 2: Behavioral Analysis...")
markov_df = run_markov_analysis(PATH_DATA, PAIR_IDS)

print("Step 3: Neural Decoding...")
decoding_data = run_decoding_analysis(epochs_dict, PATH_DATA, PAIR_IDS)

print("Step 4: Bayesian Stats...")
log_bfs = calculate_bfs(decoding_data)
```
